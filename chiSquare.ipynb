```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
df
```

```python
contingency_table=pd.crosstab(df['sex'],df['survived'])
```

```python
contingency_table
```

```python
chi2,p_value,dof,expected=chi2_contingency(contingency_table)
```

```python
expected
```

```python
dof
```

```python
p_value
```

```python
chi2
```

```python
alpha = 0.05

if p_value < alpha:
    print("I will reject the null hypothesis and there is a significat relation between gender survival")
else:
    print("there is no connection")
```
